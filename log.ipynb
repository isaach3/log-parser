```python
csvs = [f for f in os.listdir() if '.csv' in f.lower()]

os.mkdir('csv_files')

for csv in csvs:
    df = pd.read_csv(csv,encoding='utf-8')
    df['Datetime'] = pd.to_datetime(df['Datetime'],format = '[%d/%b/%Y:%H:%M:%S%z]')
    df.rename(columns={'Unknown Client ID':'Unknown'},inplace=True)
    df.to_csv(csv,encoding='utf-8')
    new_path = 'csv_files/' + csv
    shutil.move(csv, new_path)
```

```python
for csv in csvs:
    os.remove(csv)
```

```python
keys = ['IP','Remote Log Name','User Identifier','Datetime','Request','Requested URL','Status Code','Size','Referrer','User Agent']
with open('access_log.2208-2302.csv','w',encoding='utf-8') as f:
    dict_writer = csv.DictWriter(f,keys)
    dict_writer.writeheader() 
```
